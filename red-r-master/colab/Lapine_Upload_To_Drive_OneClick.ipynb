# Lapine → R_Master/reference 上传接口

这个 notebook **只负责上传 Lapine 包到 Google Drive**，不跑 Blender、不接头。

目标位置：

`MyDrive/R_Master/reference/Lapine_Ver.1.11_A.zip`

上传成功后，回到 v8a Fix1 直接“全部运行”即可。


```python
from google.colab import drive, files
from pathlib import Path
import shutil, hashlib

print("Lapine 专用上传接口")
drive.mount("/content/drive")

DST = Path("/content/drive/MyDrive/R_Master/reference/Lapine_Ver.1.11_A.zip")
DST.parent.mkdir(parents=True, exist_ok=True)

if DST.exists() and DST.stat().st_size > 70*1024*1024:
    print(f"✓ 已存在：{DST}")
    print(f"✓ 大小：{DST.stat().st_size/1024/1024:.1f} MiB")
else:
    print("请选择 Lapine_Ver.1.11_A.zip")
    up = files.upload()
    if not up:
        raise RuntimeError("没有选择文件")
    name = next(iter(up))
    src = Path("/content") / name
    src.write_bytes(up[name])
    shutil.copy2(src, DST)
    print("✓ 已复制到 Google Drive")

h = hashlib.sha256()
with DST.open("rb") as f:
    for b in iter(lambda: f.read(8*1024*1024), b""):
        h.update(b)
sha = h.hexdigest()
print("SHA256:", sha)

EXPECTED="8e215aeae8d2e42719305d66292b84c2714f616b7952591dee0383c376c86af4"
if sha != EXPECTED:
    raise RuntimeError("文件校验不一致：不是之前验证过的 Lapine_Ver.1.11_A.zip")
print("✅ 上传完成且校验通过。现在可以关闭此页，回 v8a Fix1 运行。")
```
